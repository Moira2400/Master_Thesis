In this notebook I merge the all the speeches into a single, combined csv.

```python
import glob
import os
import pandas as pd
```

```python
speeches_files = glob.glob('raw_data/final_speeches_*_mother_and_baby_homes_speeches.csv')
```

```python
# Read and combine all CSVs
df_speeches_list = []
for filename in speeches_files:
    df = pd.read_csv(filename)
    df_speeches_list.append(df)
```

```python
# Concatenate all DataFrames into one
df_all_speeches = pd.concat(df_speeches_list, ignore_index=True)
```

```python
output_speeches_file = "data/merged_speeches_all_mother_and_baby_homes_speeches.csv"

df_all_speeches.to_csv(output_speeches_file, index=False)

print(f"Combined CSV saved as: {output_speeches_file}")
```
